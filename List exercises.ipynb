```python
import random
beg, end = 1, 6
summit = 0
thrownDiceNumbers = []

for i in range(5):
    roll = random.randint(beg, end)
    thrownDiceNumbers.append(roll)
    summit += roll

print("Thrown dice number: ", thrownDiceNumbers)
print("the sum is", summit)




```
