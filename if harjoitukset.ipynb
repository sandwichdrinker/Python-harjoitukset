```python
num1, num2 = 7, 7

if num1 == num2:
    print("Number 1 is equal to number 2.")
else:
    print("Number 1 is not equal to number 2.")
if num1 > num2:
    print("Number 1 is greater than number 2.")
else:
    print("Number 1 is not greater than number 2.")
if num1 >= num2:
    print("Number 1 is greater or equal to number 2.")
else:
    print("Number 1 is not greater or equal to number 2.")
```

```python
num1, num2, num3 = 7, 9, 7

if num1 == num2 and num1 == num3 and num2 == num3:
    print("All numbers are equal.")
elif num1 == num2:
    print("Number 1 is equal to number 2.")
elif num2 == num3:
    print("Number 2 is equal to number 3.")
else:
    print("None of the numbers are equal to each other.")

if num1 > num2 and num1 > num3:
    print("Number 1 is greater than numbers 2 and 3.")
elif num1 > num2:
    print("Number 1 is greater than numbers 2.")
elif num2 > num3:
    print("Number 2 is greater than number 3.")
else:
    print("None of the numbers can be measured.")

if num1 == num2:
    print("Number 1 is equal to number 2.")
elif num1 == num3:
    print("Number 1 is equal to number 3.")
else:
    print("Number 1 is not equal to either 2 or 3.")
```
