```python
correct_name = "Burnice"
attempts = 0

while True:
    guess = input("Guess the name: ")
    
    attempts += 1

    if guess == correct_name:
        print(f"Congratulations! You guessed it in {attempts} {'guess' if attempts == 1 else 'guesses'}.")
        break
    else:
        print(f"Sorry, that's incorrect! Here's a hint: {correct_name[:attempts]}")

        quit_response = input("Do you want to quit? (yes/no): ").lower()
        if quit_response == "yes" or quit_response == "y":
            print("Goodbye!")
            break 
        else:
            print("Try again!")
```
